````python
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
````

````python
iris = load_iris()
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
````

````python
print(df.head())
````

````python
df["Species"] = iris.target_names[iris.target]
````

````python
df.loc[3, "sepal length (cm)"] = np.nan
df.loc[7, "petal width (cm)"] = np.nan
df.loc[7, "sepal width (cm)"] = np.nan
df.loc[7, "petal length (cm)"] = np.nan
df.loc[7, "Species"] = np.nan
````

````python
df.isnull().sum()
````

````python
df.tail(10)
````

````python
df = df.dropna(axis=1, how='all')
````

````python
df.tail()
````

````python
df.fillna(df.mean(numeric_only=True),inplace=True)
````

````python
df.isnull().sum()
````

````python
df['sepal length (cm)'].unique()
````

````python
# the target column can't be mean/max-filled like the numeric features,
# so any row still missing a Species label is dropped before encoding
df = df.dropna(subset=["Species"])

df.isnull().sum()
````

````python
encoder = LabelEncoder()

df["Species"] = encoder.fit_transform(df['Species'])

print(df.head())
````

````python
df.fillna(df.max(numeric_only=True),inplace=True)
````

````python
df['Species'].unique()
````

````python
from sklearn.model_selection import train_test_split

X = df.iloc[:, 0:4]
y = df["Species"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
````

````python
from sklearn import svm
from sklearn.metrics import accuracy_score

classifier = svm.SVC(kernel='linear')

classifier.fit(X_train, y_train)
````

````python
X_test_prediction = classifier.predict(X_test)

testing_data_accuracy = accuracy_score(X_test_prediction, y_test)
````

````python
print("Accuracy score of the testing data: ",testing_data_accuracy)
````

````python
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, y_train)

print("Accuracy score of the training data: ", training_data_accuracy)
````

## Detailed Model Evaluation

````python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(
    y_test,
    X_test_prediction,
    target_names=encoder.classes_
))
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, X_test_prediction)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=encoder.classes_,
    yticklabels=encoder.classes_
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
````

## Create Deployment Files for Render

The cells below write out a small Flask API (`app.py`), its `requirements.txt`, and a `render.yaml` blueprint. These, together with `iris_model.pkl` and `label_encoder.pkl`, are everything Render needs to serve the model.

````python
%%writefile app.py
from flask import Flask, request, jsonify
import joblib
import pandas as pd

app = Flask(__name__)

classifier = joblib.load("iris_model.pkl")
encoder = joblib.load("label_encoder.pkl")

FEATURE_NAMES = [
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]


@app.route("/", methods=["GET"])
def home():
    return jsonify({"message": "Iris classifier API is running"})


@app.route("/predict", methods=["POST"])
def predict():
    data = request.get_json()

    try:
        input_data = [[
            data["sepal_length"],
            data["sepal_width"],
            data["petal_length"],
            data["petal_width"]
        ]]
    except KeyError as e:
        return jsonify({"error": f"Missing field: {e}"}), 400

    input_df = pd.DataFrame(input_data, columns=FEATURE_NAMES)
    prediction = classifier.predict(input_df)
    species = encoder.inverse_transform(prediction)[0]

    return jsonify({"prediction": species})


if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)
````

## Deploying to Render

1. Push `app.py`, `requirements.txt`, `render.yaml`, `iris_model.pkl`, and `label_encoder.pkl` to a GitHub repository (or unzip `deployment_bundle.zip` into one).
2. On [Render](https://render.com), choose **New +  Web Service** and connect that repository.
3. Render reads `render.yaml` automatically and sets the build command to `pip install -r requirements.txt` and the start command to `gunicorn app:app`.
4. Once deployed, send a POST request to `https://<your-service>.onrender.com/predict` with a JSON body such as:

```json
{
  "sepal_length": 5.1,
  "sepal_width": 3.5,
  "petal_length": 1.4,
  "petal_width": 0.2
}
```
